```python
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import pandas as pd
```

```python
pio.templates.default = "plotly_white"
```

```python
business = pd.read_csv(' plotly_dv_lab_3.csv')
```

```python
business.info()
```

```python
business["Date"] = pd.to_datetime(business["Date"])
```

## 1. Line Chart – Revenue Over Time

### Objective
Visualize the trend of **Revenue** across different dates.

### Business Insight
- Helps identify increasing or decreasing revenue trends.
- Detects seasonal patterns and unusual fluctuations.
- Useful for monitoring business growth over time.

```python
fig = px.line(business,x="Date",y="Revenue",title="Revenue Trend Over Time",markers=True)
fig.update_layout(xaxis_title="Date",yaxis_title="Revenue ($)")
fig.show()
```


## 2. Bar Chart – Total Revenue by Region

### Objective
Compare the **total Revenue** generated by each **Region**.

### Business Insight
- Identifies the highest and lowest performing regions.
- Makes regional performance comparisons simple.
- Assists in making region-specific business decisions.

```python
region_revenue = (business.groupby("Region", as_index=False)["Revenue"].sum())
fig = px.bar(region_revenue,x="Region",y="Revenue",color="Region",text_auto=".2s",title="Total Revenue by Region")
fig.update_layout(xaxis_title="Region",yaxis_title="Total Revenue ($)")
fig.show()
```


## 3. Scatter Plot – Marketing Spend vs Revenue

### Objective
Analyze the relationship between **Marketing Spend** and **Revenue**.

### Business Insight
- Evaluates whether increased marketing investment leads to higher revenue.
- Reveals correlations between marketing expenditure and sales.
- Highlights outliers that may require further investigation.

```python
fig = px.scatter(business,x="Marketing_Spend",y="Revenue",color="Region",hover_data=["Product_Category"],title="Marketing Spend vs Revenue")
fig.update_layout(xaxis_title="Marketing Spend ($)",yaxis_title="Revenue ($)")
fig.show()
```


## 4. Pie Chart – Product Category Distribution

### Objective
Display the proportion of records belonging to each **Product Category**.

### Business Insight
- Shows the composition of the dataset.
- Identifies dominant and less represented product categories.
- Helps understand the overall product mix.

```python
category_count = (business.groupby("Product_Category").size().reset_index(name="Count"))
fig = px.pie(category_count,names="Product_Category",values="Count",title="Distribution of Product Categories",hole=0.35)
fig.update_traces(textposition="inside", textinfo="percent+label")
fig.show()
```

## 5. Histogram – Customer Satisfaction

### Objective
Visualize the distribution of **Customer Satisfaction** scores using **15 bins**.

### Business Insight
- Shows the frequency of customer satisfaction ratings.
- Identifies the most common satisfaction levels.
- Helps evaluate overall customer experience and consistency.

```python
fig = px.histogram(business,x="Customer_Satisfaction",nbins=15,title="Distribution of Customer Satisfaction Scores")
fig.update_layout(xaxis_title="Customer Satisfaction",yaxis_title="Frequency")
fig.show()
```

## 6. Grouped Bar Chart – Average Revenue and Expenses by Region

### Objective
Compare the **average Revenue** and **average Expenses** for each **Region** using a grouped bar chart.

### Business Insight
- Compares revenue generation against operational expenses across regions.
- Identifies regions with better financial efficiency.
- Helps evaluate cost management and profitability.

```python
region_avg = (business.groupby("Region")[["Revenue", "Expenses"]].mean().reset_index())
fig = go.Figure()
fig.add_trace(go.Bar(x=region_avg["Region"],y=region_avg["Revenue"],name="Average Revenue"))
fig.add_trace(go.Bar(x=region_avg["Region"],y=region_avg["Expenses"],name="Average Expenses"))
fig.update_layout(title="Average Revenue and Expenses by Region",xaxis_title="Region",yaxis_title="Average Amount ($)",barmode="group")
fig.show()
```


## 7. Box Plot – Profit Distribution by Product Category

### Objective
Visualize the distribution of **Profit** across different **Product Categories**.

### Business Insight
- Displays the median profit for each product category.
- Identifies variations in profitability.
- Detects outliers that may represent exceptional gains or losses.

```python
fig = px.box(business,x="Product_Category",y="Profit",color="Product_Category",title="Profit Distribution by Product Category",points="outliers")
fig.update_layout(xaxis_title="Product Category",yaxis_title="Profit ($)")
fig.show()
```



## 8. Bubble Chart – Marketing Spend vs Revenue

### Objective
Enhance the scatter plot by representing **Active Users** through bubble size and **Region** through color.

### Business Insight
- Examines the relationship between marketing investment and revenue.
- Highlights customer engagement through bubble sizes.
- Enables comparison of regional performance within a single visualization.

```python
fig = px.scatter(business,x="Marketing_Spend",y="Revenue",size="Active_Users",color="Region",hover_data=["Product_Category"],
                 title="Marketing Spend vs Revenue (Bubble Chart)",size_max=40)
fig.update_layout(xaxis_title="Marketing Spend ($)",yaxis_title="Revenue ($)")
fig.show()
```

## 9. Area Chart – Revenue and Expenses Over Time

### Objective
Visualize the progression of **Revenue** and **Expenses** over time using an overlapping or stacked area chart.

### Business Insight
- Illustrates how revenue and expenses change over time.
- Makes it easy to compare financial growth with operational costs.
- Helps identify periods of increasing profitability or rising expenses.

```python
area_df = business.sort_values("Date")
fig = px.area(area_df,x="Date",y=["Revenue", "Expenses"],title="Revenue and Expenses Over Time")
fig.update_layout(xaxis_title="Date",yaxis_title="Amount ($)")
fig.show()
```


## 10. Violin Plot – Conversion Rate by Region

### Objective
Visualize the distribution and density of **Conversion Rate** across different **Regions**.

### Business Insight
- Shows the spread and concentration of conversion rates.
- Reveals differences in customer acquisition efficiency among regions.
- Helps identify regions with more consistent conversion performance.

```python
fig = px.violin(business,x="Region",y="Conversion_Rate",color="Region",box=True,points="all",title="Conversion Rate Distribution by Region")
fig.update_layout(xaxis_title="Region",yaxis_title="Conversion Rate")
fig.show()
```

## 11. Sunburst Chart – Revenue by Region and Product Category

### Objective
Create a hierarchical visualization that displays **Revenue** first by **Region** and then by **Product Category**.

### Business Insight
- Reveals the contribution of each region to the overall revenue.
- Shows which product categories perform best within each region.
- Helps identify high-performing and low-performing business segments.

```python
sunburst_df = (business.groupby(["Region", "Product_Category"], as_index=False)["Revenue"].sum())
fig = px.sunburst(sunburst_df,path=["Region", "Product_Category"],values="Revenue",title="Revenue Distribution by Region and Product Category")
fig.show()
```


## 12. 3D Scatter Plot – Marketing Spend, Revenue, and Profit

### Objective
Visualize **Marketing Spend**, **Revenue**, and **Profit** in a three-dimensional space while using **Product Category** as the color dimension.

### Business Insight
- Examines the relationship between marketing investment, revenue generation, and profitability.
- Helps identify product categories with strong financial performance.
- Detects clusters and outliers that may require further investigation.

```python
fig = px.scatter_3d(business,x="Marketing_Spend",y="Revenue",z="Profit",color="Product_Category",hover_data=["Region"],
                    title="3D Scatter Plot: Marketing Spend vs Revenue vs Profit")
fig.show()
```


## 13. Density Heatmap – Active Users vs New Signups

### Objective
Create a two-dimensional density heatmap to visualize the concentration of observations between **Active Users** and **New Signups**.

### Business Insight
- Highlights regions where customer activity is most concentrated.
- Reveals common user acquisition patterns.
- Makes it easier to identify unusual business days with exceptionally high or low engagement.

```python
fig = px.density_heatmap(business,x="Active_Users",y="New_Signups",nbinsx=25,nbinsy=25,color_continuous_scale="Viridis",
                         title="Density Heatmap: Active Users vs New Signups")
fig.show()
```


## 14. Facet Grid (Subplots) – Profit Trend by Region

### Objective
Generate separate line charts of **Profit over Time** for each **Region** using faceted subplots.

### Business Insight
- Enables comparison of regional profit trends independently.
- Highlights seasonal variations and long-term growth patterns.
- Assists in identifying consistently profitable or underperforming regions.

```python
fig = px.line(business,x="Date",y="Profit",color="Region",facet_col="Region",title="Profit Trend by Region",height=900)
fig.for_each_annotation(lambda a: a.update(text=a.text.split("=")[-1]))
fig.update_yaxes(matches=None)
fig.show()
```


## 15. Animated Scatter Plot – Monthly Marketing Performance

### Objective
Create an animated scatter plot of **Marketing Spend** versus **Revenue**, using **Month** as the animation frame and **Discount Applied** as the color variable.

### Business Insight
- Demonstrates how marketing performance changes over different months.
- Helps evaluate the impact of promotional discounts on revenue.
- Provides an interactive view of changing business trends over time.

```python
months=["January","February","March","April","May","June","July","August","September","October","November","December"]
business["Month"]=pd.Categorical(business["Month"],categories=months,ordered=True)
fig = px.scatter(business,x="Marketing_Spend",y="Revenue",animation_frame="Month",color="Discount_Applied",hover_name="Product_Category",size="Active_Users",
                 range_x=(0,business['Marketing_Spend'].max()*1.1),range_y=(0,business['Revenue'].max()*1.1),title="Marketing Spend vs Revenue Over Months")
fig.show()
```
