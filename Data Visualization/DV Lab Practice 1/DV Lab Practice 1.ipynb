```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv(" DV_Lab1_dataset.csv")
```

# Task 1: Bar Chart (Comparison)

## Objective
Compare the total **Sales_Revenue** across the four different **Regions**.

## Instructions
1. Group the dataset by the **Region** column.
2. Calculate the **sum of Sales_Revenue** for each region.
3. Create a **vertical bar chart** to display the total sales revenue by region.
4. Add a **chart title** that clearly describes the visualization.
5. Label the **X-axis** as **Region** and the **Y-axis** as **Total Sales Revenue**.

```python
sales_by_region = df.groupby("Region")["Sales_Revenue"].sum()
plt.bar(sales_by_region.index,sales_by_region.values,color='skyblue',edgecolor="black",width=0.5)
plt.xlabel('Region')
plt.ylabel('Total Sales Revenue')
plt.title('Total Sales Revenue by Region')
```

## Task 2: Line Chart (Trend over Time)

### Objective
Track the monthly trend of **Sales_Revenue** using a line chart.

### Instructions
1. Convert the **`Date`** column to the **datetime** format.
2. Extract the **month** from the `Date` column.
3. Group the dataset by **Month** and calculate the **total Sales_Revenue** for each month.
4. Plot a **line chart** with:
   - **Month** on the x-axis.
   - **Total Sales_Revenue** on the y-axis.
   - **Markers** at each data point to highlight monthly values.
5. Add an appropriate **title**, **axis labels**, and **grid** to improve readability.

```python
df["Date"] = pd.to_datetime(df["Date"])
df["Month"] = df["Date"].dt.month
monthly_sales = df.groupby("Month")["Sales_Revenue"].sum()
plt.plot(monthly_sales.index,monthly_sales.values,marker='o',linestyle='-',color='blue')
plt.xlabel('Month')
plt.ylabel('Total Sales Revenue')
plt.title('Monthly Sales Revenue Trend')
plt.grid(True)
```

## Task 3: Pie Chart (Composition)

### Objective
Visualize the market share of each **Product_Category** based on the total **Sales_Revenue**.

### Instructions
1. Group the dataset by **`Product_Category`**.
2. Calculate the **total Sales_Revenue** for each product category.
3. Plot a **pie chart** to represent the revenue contribution of each category.
4. Display the percentage share of each category using the **`autopct`** parameter.
5. Highlight the category with the **highest Sales_Revenue** by exploding its slice.
6. Add an appropriate **title** to the chart for better presentation.

```python
category_sales = df.groupby("Product_Category")["Sales_Revenue"].sum()
explode = [0.1 if value == category_sales.max() else 0 for value in category_sales]
plt.figure(figsize=(8, 8))
plt.pie(category_sales,labels=category_sales.index,autopct='%1.1f%%',startangle=90,explode=explode)
plt.title("Market Share of Product Categories")
plt.axis('equal')
```

## Task 7: Heatmap (Correlation)

### Objective
Identify which numerical features are most strongly correlated with each other.

### Instructions
1. Select the following numerical columns:
   - `Marketing_Spend`
   - `Sales_Revenue`
   - `Customer_Age`
   - `Store_Traffic`

## Task 8: Bubble Chart (Multi-dimensional)

### Objective
Visualize the relationship between **Marketing_Spend**, **Sales_Revenue**, and **Store_Traffic** simultaneously.

### Instructions
1. Create a **scatter plot**.
2. Set:
   - **Marketing_Spend** on the x-axis.

## Task 9: Area Chart (Cumulative Trend)

### Objective
Visualize the cumulative **Sales_Revenue** over time for each **Product_Category** using a stacked area chart.

### Instructions
1. Convert the **`Date`** column to the **datetime** format.
2. Create a **pivot table** with:
   - `Date` as the index.
   - `Product_Category` as the columns.
   - `Sales_Revenue` as the values.
3. Fill any missing values with **0**.
4. Use **`plt.stackplot()`** to create a stacked area chart showing the contribution of each product category over time.
5. Add an appropriate **title**, **axis labels**, and **legend** for better readability.

```python
df["Date"] = pd.to_datetime(df["Date"])
pivot_df = df.pivot_table(index="Date",columns="Product_Category",values="Sales_Revenue",aggfunc="sum",fill_value=0)
plt.figure(figsize=(10, 6))
plt.stackplot(pivot_df.index,pivot_df.T.values,labels=pivot_df.columns)
plt.title("Cumulative Sales Revenue by Product Category Over Time")
plt.xlabel("Date")
plt.ylabel("Sales Revenue")
plt.legend(title="Product Category")
plt.grid(True)
```
